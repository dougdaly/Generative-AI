````python
# Generate markdown tree
from IPython.display import Markdown, display
from pathlib import Path

def tree_markdown(root: str | Path, max_depth=6, ignore=("__pycache__", ".DS_Store", ".git")) -> str:
    root = Path(root)
    lines = [f"```\n{root.name}/"]
    def walk(p: Path, prefix: str, depth: int):
        if depth > max_depth: 
            return
        children = sorted([c for c in p.iterdir() if c.name not in ignore], key=lambda x: (x.is_file(), x.name.lower()))
        for i, c in enumerate(children):
            is_last = i == len(children)-1
            branch = "└── " if is_last else "├── "
            lines.append(prefix + branch + c.name)
            if c.is_dir():
                extension = "    " if is_last else "│   "
                walk(c, prefix + extension, depth + 1)
    walk(root, "", 1)
    lines.append("```")
    return "\n".join(lines)

md = tree_markdown("./assets/agentic-multimodal/src")
#display(Markdown(md))
print(md)
````
